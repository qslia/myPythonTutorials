```python
import math

# 计算两个数的最大公约数
result = math.gcd(48, 18)
print(result)  # 输出: 6
```

```python
def gcd_euclidean(a, b):
    while b != 0:
        a, b = b, a % b  # 同时赋值
    return abs(a)  # 取绝对值保证正数

print(gcd_euclidean(48, 18))  # 输出: 6
print(gcd_euclidean(-48, 18)) # 输出: 6

# 步骤拆解
# temp_tuple = (b, a % b)  # 先创建元组，使用旧的 a, b 值
# a = temp_tuple[0]        # 第一个值赋给 a
# b = temp_tuple[1]        # 第二个值赋给 b
```

```python
def gcd_euclidean(a, b):
    while b != 0:
        a = b 
        b = a % b  
            
    return abs(a)

print(gcd_euclidean(48, 18))  # 输出: 6
print(gcd_euclidean(-48, 18)) # 输出: 6
```

```python
a = 48
b = 18

# 执行前
print(f"执行前: a={a}, b={b}")  # a=48, b=18

# 同时赋值
a, b = b, a % b

# 执行后
print(f"执行后: a={a}, b={b}")  # a=18, b=12
```

```python
a = 48
b = 18

# 执行前
print(f"执行前: a={a}, b={b}")  # a=48, b=18

remainder = a % b
a = b
b = remainder

# 执行后
print(f"执行后: a={a}, b={b}")  # a=18, b=12
```

```python
def gcd_euclidean(a, b):
    while b != 0:
        remainder = a % b  # 先保存余数
        a = b              # 更新 a
        b = remainder      # 更新 b
    return abs(a)


print(gcd_euclidean(48, 18))  # 输出: 6
print(gcd_euclidean(-48, 18)) # 输出: 6
```
